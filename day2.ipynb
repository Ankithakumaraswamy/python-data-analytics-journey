```python
movie_name = "Jailer 2"
ticket_price = 150
print("Movie:", movie_name)

no_of_seats = int(input("Enter the number of seats: "))
result = ticket_price * no_of_seats

# Fix 1: convert the number to text with str()
print("Your ticket price is: " + str(result) + " rps")

# Fix 2: use commas, so print() handles the types for you
print("Your ticket price is:", result, "rps")
```
