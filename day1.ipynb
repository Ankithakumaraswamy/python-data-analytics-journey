```python
#Naming style	

#PascalCase	:- Every word starts with a capital letter.	   
#EX:- StudentDetails

#snake_case :- Words are separated by underscores (_).    
#EX:- student_details

#camelCase :- First word starts lowercase; 
#subsequent words start with capitals.	 
#EX:- studentDetails


MyName = "ankitha" #pascal case 
my_name = "ankitha" #snake case (preferred by python)
myName = "ankitha" #camel case
print(MyName, my_name, myName)
```
