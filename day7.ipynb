```python
names = ["ankitha", "anvi", "dinosaur", "vishank", "aarvi", "viyank"]   #---> list
#[start:stop:step]
print(names[1:4])
```

```python
names = ["ankitha", "anvi", "dinosaur", "vishank", "aarvi", "viyank"]
#[start:stop:step]
print(names[0:4])
```

```python
names = ["ankitha", "anvi", "dinosaur", "vishank", "aarvi", "viyank"]
#[start:stop:step]
print(names[1:4])
print(names[3:6])
print(names[-1]) #----> negative indexing
print(names[0:5:2])
print(names[1:6:2])
```

```python
names = ["ankitha", "anvi", "dinosaur", "vishank", "aarvi", "viyank"]
#to find last before value (even if you have 1000 values, the logic for last before value will be same)
print(names[-2])
print(names[::-1]) #----> reverse the list
print(names[::-2]) #----> reverse the list with step 2
#when given negative index, all values must be in negative only
```
