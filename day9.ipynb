```python
#Dictionary
#key value pair, uses : btw them
#mutable, keys cannot be duplicated, values can
details = { "name" : "Ankitha",
           "age" : 20,
          "designation" : "Data Analyst"}
print(details)
```

```python
details = { "name" : "Ankitha",
           "age" : 20,
          "designation" : "Data Analyst"}
print(details['name'])
```

```python
details = { "name" : "Ankitha",
           "age" : 20,
          "designation" : "Data Analyst"}
print(details)
details['name'] = "Anvi"
print(details['name'])
```

```python
details = { "name" : "Ankitha",
           "age" : 20,
          "designation" : "Data Analyst",
            "name" : "Vishank"}
print(details)
details['name'] = "Anvi"
print(details['name'])
```

```python
details = { "name" : "Ankitha",
           "age" : 20,
          "designation" : "Data Analyst",
            "name2" : "Ankitha"}
print(details)
```

```python
#break, continue, pass
#pass - do nothing

a = "a,b,c,d"
print(a)
```

```python
a="a,b,c,d"
for i in a:
  print(i)
```

```python
a="a,b,c,d"
for i in a:
  print(i, end = "")
```

```python
a="a,b,c,d"
for i in a:
  if i == ",":
    continue
  else:
    print(i, end = " ")
```

```python
a="a,b,c,d"
for i in a:
  if i == ",":
    continue
    print("Hii")
  else:
    print(i, end = " ")
```
