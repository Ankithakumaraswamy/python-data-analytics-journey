```python
#2dlist - list inside list
ka = ["bengaluru","mysuru","mandya"]
tn = ["chennai","coimbatore","madurai"]
ap = ["hyderabad","vijayawada","visakhapatnam"]

states = [ka, tn, ap]
print(states) 
```

```python
#memory allocation
#stack and heap memory 
ka = ["bengaluru","mysuru","mandya"]
tn = ["chennai","coimbatore","madurai"]
ap = ["hyderabad","vijayawada","visakhapatnam"]

states = [ka, tn, ap]
indianstates = states
indianstates.remove(tn) #it will remove tn from both states and indianstates because both are pointing to same memory location
#it shallow copies and is stored in head memory.
print(states)
print(indianstates)
```

```python
#memory management
#copy (shallowcopy and deepcopy)
import copy
ka = ["bengaluru","mysuru","mandya"]
tn = ["chennai","coimbatore","madurai"]
ap = ["hyderabad","vijayawada","visakhapatnam"]

states = [ka, tn, ap]
indianstates= copy.deepcopy(states)
#this will create a new memory location for indianstates and it will not affect the states list when we remove tn from indianstates
#this is deep copies and is stored in stack memory.
indianstates.remove(tn)
print(states)
print(indianstates)
```

```python
#split and join
#split is used to convert a string into a list
#join is used to convert a list into a string 
names = "abc,def,ghi,jkl"
splittednames = names.split(",")
print(splittednames)
joinednames = "-".join(splittednames)
print(joinednames)
```

```python
names = "abc,def,ghi,jkl"
splittednames = names.split("g")
print(splittednames)
joinednames = "%".join(splittednames)
print(joinednames)
```

```python
#palindrome
a = input("enter a string: ")
print(a)
b = []
for i in a:
    b.append(i)
print(b)
c=b[::-1]
print(c)
d = ''.join(c)
print(d)
if a==d:
    print("it is a palindrome")
else:
    print("it is not a palindrome")
```
