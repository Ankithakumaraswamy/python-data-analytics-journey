```python
#Write a program to calculate the total shopping bill. 
#The user can buy multiple items with different prices.
#Ask the user to enter the price of each item.
#Also, prompt the user to enter the quantity of each item.
#Calculate the total cost for each item and sum them up to get the final bill.
#Display the total bill amount.

total_bill = 0
n_items = int(input("Enter the number of items you want to buy: "))

for i in range(n_items):
    price = float(input("Enter the price of item : "))
    quantity = float(input("Enter the quantity of item : "))
    total_bill += price * quantity
print("Total bill amount: ", total_bill)
```

```python
#Given the list lst = [1, 2, 3, 4, 5, 6], 
#write a loop that doubles the value of each element at even indices 
#and triples the value at odd indices.

lst = [1,2,3,4,5,6]
for i in range(len(lst)):
    if i % 2 == 0:
        lst[i] *= 2
    else:
        lst[i] *= 3
print(lst)
```

```python
#Write a Python program that 
#prints numbers from 1 to 50. 
#For multiples of 3, print "hello" instead of the number. 
#For multiples of 5, print "world." 
#For numbers which are multiples of both 3 and 5, print "helloworld." 
#Use a for loop, if-else, and elif to solve this.

#Expected Output:
#1, 2, hello, 4, world, hello, 7, 8, hello, world, 11, hello, 13, 14, helloworld, ...

for i in range(1, 51):
    if i % 3 == 0:
        print("hello", end=", ")
    elif i % 5 == 0:
        print("world", end=", ")
    elif i % 3 == 0 and i % 5 == 0:
        print("helloworld", end=", ")
    else:
        print(i, end=", ")
```

```python
#Write a Python function 
#that takes a list of numbers as input and 
#returns a new list containing only the even numbers 
#from the input list.
#I/p - [1,2,3,4,5,6,7,8]
#O/p -[2,4,6,8]

def list_even_no(ip_list):
    even_no = []
    for n in ip_list:
        if n % 2 == 0:
            even_no.append(n)
    return even_no

ip_list = [1, 2, 3, 4, 5, 6, 7, 8]
result = list_even_no(ip_list)
print("Even numbers:", result)
```

```python
#Write a program that repeatedly asks the user to input a number. 
#The program should print the square of the number. 
#The loop should continue until the user inputs 0. 
#If the user inputs a non-numeric value, print an error message 
#and ask for the input again.
#Example 
#Enter a number: 3
#Output: 9
#Enter a number: abc
#Output: Please enter a valid number.
#Enter a number: 0
#Output: Exiting the program.

while True:
    n = input("Enter a number: ")
    if n.isdigit():
        n = int(n)
        if n == 0:
            print("Exiting the program.")
            break
        else:
            print("Output:", n ** 2)
    else:
        print("Please enter a valid number")
```

```python
#Given two lists of integers, 
#write a Python program that finds 
#the intersection (common elements) of the two lists. 
#Use a for loop to iterate through the lists and 
#if-else to check for common elements.

#Input:
#List 1: [1, 2, 3, 4, 5]
#List 2: [4, 5, 6, 7, 8]
#Expected Output:
#Common elements: [4, 5]

list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
common_elements = []
for element in list1:
    if element in list2:
        common_elements.append(element)
print("Common elements:", common_elements)
```

```python
#Write a Python program that continuously takes user input
#until the user enters 0.
#The program should keep a running sum of
#positive numbers and negative numbers separately.
#Use a while loop for continuous input,
#and if-else to classify numbers as positive or negative.
#Enter a number: 5
#Enter a number: -3
#Enter a number: 8
#Enter a number: -2
#Enter a number: 0
#O/p: 
#Sum of positive numbers: 13
#Sum of negative numbers: -5

positive_sum = 0
negative_sum = 0

while True:
    n = int(input("Enter a number: "))
    if n == 0:
        break
    elif n > 0:
        positive_sum += n
    else:
        negative_sum += n

print("Sum of positive numbers:", positive_sum)
print("Sum of negative numbers:", negative_sum)
```

```python
#Write a Python program that takes a list of integers
#and separates the even and odd numbers into two different lists. 
#Use a for loop to iterate through the list, and 
#if-else to check if the number is even or odd.

#Input:
#A list of integers: [12, 7, 9, 3, 10, 5]
#Expected Output:
#Even numbers: [12, 10]
#Odd numbers: [7, 9, 3, 5]

n = [12, 7, 9, 3, 10, 5]
even_numbers = []
odd_numbers = []
for num in n:
    if num % 2 == 0:
        even_numbers.append(num)
    else:
        odd_numbers.append(num)

print("Even numbers:", even_numbers)
print("Odd numbers:", odd_numbers)
```
