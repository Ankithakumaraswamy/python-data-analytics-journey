```python
metro = "yes"
if metro == "yes":
    print("Take metro")
else:
    print("Take bus")
```

```python
gender = "female"
age = 45
if gender == "female" and age <= 45:
    print("welcome")
else:
    print("Get lost")
```

```python
gender = "female"
age = 47
if gender == "female" or age <= 45:
    print("welcome")
else:
    print("Get lost")
```

```python
metro = "yes"
bus = "no"
if metro == "yes":
    print("Take metro")
elif bus == "yes":
    print("take bus")
else:
    print("take rapido")
```

```python
metro = "no"
bus = "no"
if metro == "yes":
    print("Take metro")
elif bus == "yes":
    print("take bus")
else:
    print("take rapido")
```

```python
gender = "female"
age = 48
course = "yes"

if gender == "female" and age <= 45 or course == "yes":
  print("welcome")
else :
  print("Get lost")
```
