```python
#multiplication table generator
#2 * 1 = 2
#2 * 2 = 4...
table = int(input("Enter a number: "))
num = int(input("Enter a number: "))

for i in range(1,num+1):
    result = table * i
    print(str(table) + "*" + str(i) + "=" + str(result))
```
