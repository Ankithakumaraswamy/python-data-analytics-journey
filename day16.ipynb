```python
#map
number = [1,2,3,4,5]
square = map(lambda x : x**2, number)
print(list(square))
```

```python
#reduce
from functools import reduce
number = [1,2,3,4,5]
result = reduce(lambda x,y : x+y, number)
print(result)
```

```python
#filter
number = [1,2,3,4,5]
even = filter(lambda x : x%2 ==0, number)
odd = filter(lambda x : x%2!=0, number)
print(list(even))
print(list(odd))
```

```python
#zip
student = ["Ankitha", "Bhavana", "Anvi"]
marks = [90,69,53]

result = list(zip(student,marks))
print(result)
```

```python
#matplotlib-line chart
import matplotlib.pyplot as plt
x = [20,30,40,50]
y = [2,3,4,5]

plt.plot(x,y)
plt.xlabel("Marks")
plt.ylabel("No of hours")
plt.show()
```

```python
#matplotlib-bar chart
import matplotlib.pyplot as plt
x = [20,30,40,50]
y = [2,3,4,5]

plt.bar(x,y)
plt.xlabel("Marks")
plt.ylabel("No of hours")
plt.show()
```

```python
#matplotlib-scatter chart
import matplotlib.pyplot as plt
x = [20,30,40,50]
y = [2,3,4,5]

plt.scatter(x,y)
plt.xlabel("Marks")
plt.ylabel("No of hours")
plt.show()
```
