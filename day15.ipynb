```python
# error handling / exceptional handling - except acts like an 'if' statement 
# it is used to catch the exceptions and handle the error preventing the code from crashing
try:
  num1 = int(input("Enetr a number : "))
  num2 = int(input("Enter a number : "))
  result = num1/num2
except ValueError:
  print("Enter a valid number")
except ZeroDivisionError:
  print("Denominator cannot be zero")
else:
  print(result)
```

```python
# factorial
def fact(num):
  if num == 1:
    return 1
  return num*fact(num-1)
print(fact(5))
```

```python
# generators - it is used to returns value one at a time instead of storing it in memory
def num():
  for i in range(1,10):
    yield i # you can use yield inside a for loop without for loop you cannot use it
# num() - this is give you a object you can iterate this object
data = num()
for i in data:
  print(i)
```

```python
# lambda is used to write code in a single line
add = lambda i : i+1
print(add(1))

oe = lambda i : "even" if i%2==0 else "odd"
print(oe(8))
```
