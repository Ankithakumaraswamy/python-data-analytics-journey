```python
age = int(input("enter your age: "))
if age < 18:
    print("Children ticket")
elif age >=18 and age <=75:
    is_student = input("are you a student yes/no :").lower()
    if is_student == "yes":
        print("Student ticket")
    elif is_student == "no":
        print("Adult ticket")
    else:
        print("Invalid")
else:
    print("Senior citizen ticket")
```
