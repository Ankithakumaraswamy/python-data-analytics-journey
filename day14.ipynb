```python
#strip - used for removing extra white spaces
name = "   Ankitha   "
print(name)
print(name.strip())
print(name.lstrip()) #left strip
print(name.rstrip()) #right strip

name2 = "##@!Ankitha@!##"
print(name2.strip("#@!")) #removes the characters specified in the strip method
print(name2.lstrip("#@!")) #left strip
print(name2.rstrip("#@!")) #right strip
```
